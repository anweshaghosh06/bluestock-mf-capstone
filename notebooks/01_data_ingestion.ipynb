```python
import pandas as pd
from pathlib import Path

fm = pd.read_csv(Path("../data/raw/01_fund_master.csv"))

print("=== FUND MASTER EXPLORATION ===")
print(f"Total schemes: {len(fm)}")
print(f"\n--- Fund Houses ({fm['fund_house'].nunique()} unique) ---")
print(fm['fund_house'].value_counts().to_string())
print(f"\n--- Categories ---")
print(fm['category'].value_counts().to_string())
print(f"\n--- Sub-categories ---")
print(fm['sub_category'].value_counts().to_string())
print(f"\n--- Risk Categories ---")
print(fm['risk_category'].value_counts().to_string())
print(f"\n--- Expense Ratio range ---")
print(fm['expense_ratio_pct'].describe())
```
